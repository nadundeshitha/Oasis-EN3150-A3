# DermaMNIST Exploration

This notebook demonstrates how to load the DermaMNIST dataset at 64x64 resolution, appropriately split into 70% Train, 15% Validation, and 15% Testing.

```python
import sys
import os
sys.path.append(os.path.abspath('../src'))

import matplotlib.pyplot as plt
from dataset import load_dermamnist
```

```python
# Load the raw dataset splits (already formatted to 70/15/15 and 64x64)
train_data, val_data, test_data, info = load_dermamnist()
x_train, y_train = train_data

print("\nClass Mapping:", info['label'])
```

```python
# Visualize a 3x3 grid of the first 9 training samples
plt.figure(figsize=(8, 8))
for i in range(9):
    ax = plt.subplot(3, 3, i + 1)
    plt.imshow(x_train[i])
    class_idx = str(y_train[i])
    plt.title(info['label'][class_idx])
    plt.axis("off")
plt.tight_layout()
plt.show()
```
